```python
import numpy as np
from bsm import Call,Put
import matplotlib.pyplot as plt

#dividend yield, volatility and risk free interest rate must be entered in decimal form, not percentage form
```

```python
#time dependence

current=50
strike=46
############################
time=np.linspace(.001,1,500)
############################
risk_free_interest_rate=.08
dividend_yield=.03
volatility=.3

xinterval=.05
yinterval=.5

data=[current,strike,time,risk_free_interest_rate,dividend_yield,volatility]
cp=max(Call(*data))
pp=max(Put(*data))

plt.figure(figsize=(14,7))
plt.xticks(np.arange(0,1.1*max(time),xinterval))
plt.yticks(np.arange(0,1.1*max(cp,pp),yinterval))
plt.xlabel('Time until maturity (years)')
plt.ylabel('Option price ($)')
plt.plot(time,Call(*data),label='Call')
plt.plot(time,Put(*data),label='Put')
plt.grid(True)
plt.legend()
plt.show()
```

```python
#risk free interest rate dependence

current=47
strike=50
time=0.25
###############################################
risk_free_interest_rate=np.linspace(0,.05,500)
rfir=risk_free_interest_rate
#############################
dividend_yield=0
volatility=.2

xinterval=.25
yinterval=.25

data=[current,strike,time,risk_free_interest_rate,dividend_yield,volatility]
cp=max(Call(*data))
pp=max(Put(*data))

plt.figure(figsize=(14,7))
plt.xticks(np.arange(0,110*(max(rfir)),xinterval))
plt.yticks(np.arange(0,1.1*max(cp,pp),yinterval))
plt.xlabel('Risk free interest rate (%)')
plt.ylabel('Option price ($)')
plt.plot(100*rfir,Call(*data),label='Call')
plt.plot(100*rfir,Put(*data),label='Put')
plt.grid(True)
plt.legend()
plt.show()
```

```python
#dividend yield dependence

current=220
strike=230
time=1
risk_free_interest_rate=.05
######################################
dividend_yield=np.linspace(0,.05,500)
dy=dividend_yield
##################
volatility=.3

xinterval=.25
yinterval=1

data=[current,strike,time,risk_free_interest_rate,dividend_yield,volatility]
cp=max(Call(*data))
pp=max(Put(*data))

plt.figure(figsize=(14,7))
plt.xticks(np.arange(0,110*(max(dy)),xinterval))
plt.yticks(np.arange(0,110*(max(cp,pp)),yinterval))
plt.xlabel('Dividend yield (%)')
plt.ylabel('Option price ($)')
plt.plot(100*dy,Call(*data),label='Call')
plt.plot(100*dy,Put(*data),label='Put')
plt.grid(True)
plt.legend()
plt.show()
```
