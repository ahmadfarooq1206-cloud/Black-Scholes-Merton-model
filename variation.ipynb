```python
import greeks
import numpy as np
import matplotlib.pyplot as plt
```

```python
K=120
t=1
r=.04
q=0
o=.3

lower=round(.5*K)
upper=round(1.5*K)
S=np.linspace(lower,upper,1000)

d1=(np.log(S/K)+t*(r-q+.5*o**2))/(o*t**.5)
d2=d1-o*t**.5

call_delta_values=greeks.Call_Delta(q,t,d1)
put_delta_values=greeks.Put_Delta(q,t,d1)

x_spacing=5
y_spacing=.1

plt.figure(figsize=(14,7))
plt.title('Sensitivity to stock price variation')
plt.xticks(np.arange(round(lower),round(1+upper),x_spacing))
plt.yticks(np.arange(-1,1.1,y_spacing))
plt.xlabel('Stock prices ($)')
plt.ylabel('Delta values')
plt.plot(S,call_delta_values,label='Call Delta')
plt.plot(S,put_delta_values,label='Put Delta')
plt.legend()
plt.grid(True)
plt.show()
```

```python
gamma_values=greeks.Gamma(S,q,t,o,d1)

x_spacing=5
y_spacing=.001

plt.figure(figsize=(14,7))
plt.title('Rate of change of Delta')
plt.xticks(np.arange(lower,1+upper,x_spacing))
plt.yticks(np.arange(-1,1,y_spacing))
plt.xlabel('Stock prices ($)')
plt.ylabel('Gamma values (1/$)')
plt.plot(S,gamma_values,label='Gamma')
plt.legend()
plt.grid(True)
plt.show()
```

```python
#theta: time decay

S=50
K=55
r=.04
q=.02
o=.2

upper=2
t=np.linspace(.001,upper,1000)

d1=(np.log(S/K)+t*(r-q+.5*o**2))/(o*t**.5)
d2=d1-o*t**.5

call_theta_values=greeks.Call_Theta(S,K,t,r,q,o,d1,d2)
put_theta_values=greeks.Put_Theta(S,K,t,r,q,o,d1,d2)

x_spacing=.1
y_spacing=.001

plt.figure(figsize=(14,7))
plt.title('Options price decay')
plt.xticks(np.arange(0,upper+.1,x_spacing))
plt.yticks(np.arange(-1,.1,y_spacing))
plt.xlabel('Time (Years)')
plt.ylabel('Theta values ($/Day)')
plt.plot(t,call_theta_values,label='Call Theta')
plt.plot(t,put_theta_values,label='Put Theta')
plt.legend()
plt.grid(True)
plt.show()
```

```python
S=70
K=65
t=1
r=.03
q=0

upper=.5
o=np.linspace(.001,upper,1000)

d1=(np.log(S/K)+t*(r-q+.5*o**2))/(o*t**.5)
d2=d1-o*t**.5

vega_values=greeks.Vega(S,q,t,d1)

x_spacing=2
y_spacing=.02

plt.figure(figsize=(14,7))
plt.title('Sensitivity to volatility variation')
plt.xticks(np.arange(0,101*upper,x_spacing))
plt.yticks(np.arange(0,1,y_spacing))
plt.xlabel('Volatility (%)')
plt.ylabel('Vega values ($/%)')
plt.plot(100*o,vega_values,label='Vega')
plt.legend()
plt.grid(True)
plt.show()
```

```python
S=200
K=220
t=.5
q=.02
o=.3

upper=.1
r=np.linspace(0,upper,1000)

d1=(np.log(S/K)+t*(r-q+.5*o**2))/(o*t**.5)
d2=d1-o*t**.5

call_rho_values=greeks.Call_Rho(K,t,r,d2)
put_rho_values=greeks.Put_Rho(K,t,r,d2)

x_spacing=.5
y_spacing=.1

plt.figure(figsize=(14,7))
plt.title('Sensitivity to risk free interest rate variation')
plt.xticks(np.arange(0,101*upper,x_spacing))
plt.yticks(np.arange(-1,1,y_spacing))
plt.xlabel('Risk free interest rate (%)')
plt.ylabel('Theta values ($/%)')
plt.plot(100*r,call_rho_values,label='Call Rho')
plt.plot(100*r,put_rho_values,label='Put Rho')
plt.legend()
plt.grid()
plt.show()
```
