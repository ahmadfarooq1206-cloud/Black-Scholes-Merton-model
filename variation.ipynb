```python
import numpy as np
from bsm import Call,Put
import matplotlib.pyplot as plt

#Dividend yield, volatility and risk free interest rate must be entered in decimal form, not percentage form
```

```python
#Dependence on the time to maturity

#Fixed variables
current=50
strike=46
risk_free_interest_rate=.08
dividend_yield=.03
volatility=.3

#Independent variable
time=np.linspace(.001,1,500)

#Grid spacing
xinterval=.05
yinterval=.5

data=[current,strike,time,risk_free_interest_rate,dividend_yield,volatility]
cp=max(Call(*data))
pp=max(Put(*data))

plt.figure(figsize=(14,7))
plt.xticks(np.arange(0,1.1*max(time),xinterval))
plt.yticks(np.arange(0,1.1*max(cp,pp),yinterval))
plt.xlabel('Time until maturity (years)')
plt.ylabel('Option price ($)')
plt.plot(time,Call(*data),label='Call')
plt.plot(time,Put(*data),label='Put')
plt.grid()
plt.legend()
plt.show()
```

```python
#Dependence on the risk free interest rate

#Fixed variables
current=47
strike=50
time=0.25
dividend_yield=0
volatility=.2

#Independent variable
risk_free_interest_rate=np.linspace(0,.05,500)
rfir=risk_free_interest_rate

#Grid spacing
xinterval=.25
yinterval=.25

data=[current,strike,time,risk_free_interest_rate,dividend_yield,volatility]
cp=max(Call(*data))
pp=max(Put(*data))

plt.figure(figsize=(14,7))
plt.xticks(np.arange(0,110*(max(rfir)),xinterval))
plt.yticks(np.arange(0,1.1*max(cp,pp),yinterval))
plt.xlabel('Risk free interest rate (%)')
plt.ylabel('Option price ($)')
plt.plot(100*rfir,Call(*data),label='Call')
plt.plot(100*rfir,Put(*data),label='Put')
plt.grid()
plt.legend()
plt.show()
```
